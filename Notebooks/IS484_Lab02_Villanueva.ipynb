# Laboratorio 02
---

>Caso:  caso: un repartidor debe ir desde el Almacén hasta la Zona Norte, pudiendo pasar por distintos puntos intermedios de la ciudad.

- *Estado inicial:*   El repartidor está en el almacén.
- *Estado objetivo:*  El repartidor llega a la zona norte.
- *Acciones posibles:*   Moverse a cada nodo.
- *Costo de acción:*  El costo es el mismo para cada acción.
