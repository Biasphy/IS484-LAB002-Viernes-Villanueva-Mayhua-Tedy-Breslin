## Parte A de BFS

>Caso:  caso: un repartidor debe ir desde el Almacén hasta la Zona Norte, pudiendo pasar por distintos puntos intermedios de la ciudad.

- *Estado inicial:*   El repartidor está en el almacén.
- *Estado objetivo:*  El repartidor llega a la zona norte.
- *Acciones posibles:*   Moverse a cada nodo.
- *Costo de acción:*  El costo es el mismo para cada acción.

```python
mapa = {
    "Almacen": ["Centro", "Terminal"],
    "Centro": ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal": ["Almacen", "Mercado"],
    "Mercado": ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"],
}
print(mapa)
```

## Parte B de BFS

```python
#Definiendo el algoritmo para encontrar caminos de tipo BFS
def bfs(grafo, inicio, objetivo):
    """
    Busca el camino mas corto entre 'inicio' y 'objetivo' usando BFS.
    """
    cola = [[inicio]] # cola de caminos, empieza con el estado inicial, aquí se guardan los nodos empezando por el primero (inicio)
    visitados = set() # Definiendo los nodos visitados dentro de un tipo de datos conjunto.

    #Abrimos bucle para iterar todos los caminos
    while cola:
        camino = cola.pop(0) # sacar el primer camino de la cola (FIFO)
        nodo_actual = camino[-1] # Tomamos el último elemento para saber donde estamos.

        if nodo_actual == objetivo: # Aquí verificamos si se encontró el nodo objetivo, si es así detenemos le algoritmo.
            return camino
        
        if nodo_actual not in visitados: # Si el nodo actual no ha sido visitado se añade con `add` al conjunto para evitar bucles con este nodo, permitiendo que se busquen en los otros nodos.
            visitados.add(nodo_actual)
            for vecino in grafo[nodo_actual]: # Se inicia un recorrido para visitar todos los caminos desde el nodo actual.
                nuevo_camino = camino + [vecino] # Se crea la ruta desde el nodo actual con el vecino analizado, siendo este el que sera analizado en el futuro gracias al append.
                cola.append(nuevo_camino)
                pass

    return None # no se encontró un camino

resultado = bfs(mapa, "Almacen", "ZonaNorte") # Aplicamos el algoritmo a nuestro mapa de nodos, con Alamacen como inicio y ZonaNorte como objetivo.
print("Camino encontrado:", resultado)
```
